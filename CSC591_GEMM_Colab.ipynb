## Section 5: Create Python Wrappers for CUDA Kernels

Load the compiled library and create wrapper functions to call from Python.

````python
import ctypes
import os
import numpy as np

# Find the compiled library
build_dir = '/root/CSC-591_FinalProject/src/cuda_kernels/build'
lib_candidates = [
    os.path.join(build_dir, 'libgemm_kernels.so'),
    os.path.join(build_dir, 'gemm_kernels.dll'),
    os.path.join(build_dir, 'Release', 'gemm_kernels.dll'),
]

lib_path = None
for candidate in lib_candidates:
    if os.path.exists(candidate):
        lib_path = candidate
        break

if lib_path is None:
    print("ERROR: Could not find compiled library!")
    print(f"Searched in: {lib_candidates}")
else:
    print(f"Ã¢Å“â€œ Found library at: {lib_path}")

# Load the shared library
try:
    lib = ctypes.CDLL(lib_path)
    print(f"Ã¢Å“â€œ Successfully loaded library: {lib_path}")
except Exception as e:
    print(f"ERROR loading library: {e}")
    raise

# Define function signatures
c_float_p = ctypes.POINTER(ctypes.c_float)
c_int = ctypes.c_int

# Configure function signatures
lib.naive_gemm.argtypes = [c_float_p, c_float_p, c_float_p, c_int, c_int, c_int]
lib.naive_gemm.restype = c_int

lib.tiled_gemm.argtypes = [c_float_p, c_float_p, c_float_p, c_int, c_int, c_int]
lib.tiled_gemm.restype = c_int

lib.register_blocked_gemm.argtypes = [c_float_p, c_float_p, c_float_p, c_int, c_int, c_int]
lib.register_blocked_gemm.restype = c_int

lib.hpc_gemm.argtypes = [c_float_p, c_float_p, c_float_p, c_int, c_int, c_int]
lib.hpc_gemm.restype = c_int

lib.ultra_gemm.argtypes = [c_float_p, c_float_p, c_float_p, c_int, c_int, c_int]
lib.ultra_gemm.restype = c_int

lib.turbo_gemm.argtypes = [c_float_p, c_float_p, c_float_p, c_int, c_int, c_int]
lib.turbo_gemm.restype = c_int

print("\nÃ¢Å“â€œ All kernel functions registered")

# Helper wrapper functions
def call_naive_gemm(a, b):
    """Call naive GEMM kernel. Returns result matrix C."""
    m, k = a.shape
    k, n = b.shape
    
    # Ensure contiguous float32 arrays
    a = np.ascontiguousarray(a, dtype=np.float32)
    b = np.ascontiguousarray(b, dtype=np.float32)
    c = np.zeros((m, n), dtype=np.float32)
    
    status = lib.naive_gemm(
        a.ctypes.data_as(c_float_p),
        b.ctypes.data_as(c_float_p),
        c.ctypes.data_as(c_float_p),
        c_int(m), c_int(n), c_int(k)
    )
    
    if status != 0:
        raise RuntimeError(f"naive_gemm failed with status {status}")
    return c

def call_tiled_gemm(a, b):
    """Call tiled GEMM kernel. Returns result matrix C."""
    m, k = a.shape
    k, n = b.shape
    
    # Ensure contiguous float32 arrays
    a = np.ascontiguousarray(a, dtype=np.float32)
    b = np.ascontiguousarray(b, dtype=np.float32)
    c = np.zeros((m, n), dtype=np.float32)
    
    status = lib.tiled_gemm(
        a.ctypes.data_as(c_float_p),
        b.ctypes.data_as(c_float_p),
        c.ctypes.data_as(c_float_p),
        c_int(m), c_int(n), c_int(k)
    )
    
    if status != 0:
        raise RuntimeError(f"tiled_gemm failed with status {status}")
    return c

def call_register_blocked_gemm(a, b):
    """Call register-blocked GEMM kernel. Returns result matrix C."""
    m, k = a.shape
    k, n = b.shape
    
    # Ensure contiguous float32 arrays
    a = np.ascontiguousarray(a, dtype=np.float32)
    b = np.ascontiguousarray(b, dtype=np.float32)
    c = np.zeros((m, n), dtype=np.float32)
    
    status = lib.register_blocked_gemm(
        a.ctypes.data_as(c_float_p),
        b.ctypes.data_as(c_float_p),
        c.ctypes.data_as(c_float_p),
        c_int(m), c_int(n), c_int(k)
    )
    
    if status != 0:
        raise RuntimeError(f"register_blocked_gemm failed with status {status}")
    return c

def call_hpc_gemm(a, b):
    m, k = a.shape
    k, n = b.shape
    a = np.ascontiguousarray(a, dtype=np.float32)
    b = np.ascontiguousarray(b, dtype=np.float32)
    c = np.zeros((m, n), dtype=np.float32)
    status = lib.hpc_gemm(
        a.ctypes.data_as(c_float_p),
        b.ctypes.data_as(c_float_p),
        c.ctypes.data_as(c_float_p),
        c_int(m), c_int(n), c_int(k)
    )
    if status != 0:
        raise RuntimeError(f"hpc_gemm failed with status {status}")
    return c

def call_ultra_gemm(a, b):
    m, k = a.shape
    k, n = b.shape
    a = np.ascontiguousarray(a, dtype=np.float32)
    b = np.ascontiguousarray(b, dtype=np.float32)
    c = np.zeros((m, n), dtype=np.float32)
    status = lib.ultra_gemm(
        a.ctypes.data_as(c_float_p),
        b.ctypes.data_as(c_float_p),
        c.ctypes.data_as(c_float_p),
        c_int(m), c_int(n), c_int(k)
    )
    if status != 0:
        raise RuntimeError(f"ultra_gemm failed with status {status}")
    return c

def call_turbo_gemm(a, b):
    m, k = a.shape
    k, n = b.shape
    a = np.ascontiguousarray(a, dtype=np.float32)
    b = np.ascontiguousarray(b, dtype=np.float32)
    c = np.zeros((m, n), dtype=np.float32)
    status = lib.turbo_gemm(
        a.ctypes.data_as(c_float_p),
        b.ctypes.data_as(c_float_p),
        c.ctypes.data_as(c_float_p),
        c_int(m), c_int(n), c_int(k)
    )
    if status != 0:
        raise RuntimeError(f"turbo_gemm failed with status {status}")
    return c

print("All wrapper functions created successfully!")
````

## Section 6: Correctness Testing

Validate that the CUDA kernels produce correct results by comparing against NumPy and PyTorch.

````python
import numpy as np
import torch

def check_correctness(name, cuda_result, reference, tol=1e-4):
    """Compare CUDA result with reference using multiple metrics."""
    if cuda_result.shape != reference.shape:
        print(f"  Ã¢Å“â€” Shape mismatch: {cuda_result.shape} vs {reference.shape}")
        return False
    
    # Convert to float64 for better precision comparison
    cuda_f64 = cuda_result.astype(np.float64)
    ref_f64 = reference.astype(np.float64)
    
    # Compute error metrics
    abs_error = np.abs(cuda_f64 - ref_f64)
    max_abs_error = np.max(abs_error)
    mean_abs_error = np.mean(abs_error)
    
    # Check if relative error is small
    rel_error = np.abs(cuda_f64 - ref_f64) / (np.abs(ref_f64) + 1e-8)
    max_rel_error = np.max(rel_error)
    
    passed = max_abs_error < tol
    status = "Ã¢Å“â€œ" if passed else "Ã¢Å“â€”"
    
    print(f"  {status} {name}:")
    print(f"    Max Absolute Error: {max_abs_error:.2e}")
    print(f"    Mean Absolute Error: {mean_abs_error:.2e}")
    print(f"    Max Relative Error: {max_rel_error:.2e}")
    
    return passed

# Test with small matrices first
print("="*60)
print("CORRECTNESS TEST: Small Matrices (64x64x64)")
print("="*60)

np.random.seed(42)
m, n, k = 64, 64, 64

A = np.random.randn(m, k).astype(np.float32)
B = np.random.randn(k, n).astype(np.float32)

# Reference result using NumPy
ref_result = np.dot(A, B).astype(np.float32)

# Test each kernel
kernels = [
    ("Naive GEMM", call_naive_gemm),
    ("Tiled GEMM", call_tiled_gemm),
    ("Register-Blocked GEMM", call_register_blocked_gemm),
    ("HPC GEMM (128x128, 8x8)", call_hpc_gemm),
    ("Ultra GEMM (128x128, K=32)", call_ultra_gemm),
    ("Turbo GEMM (interleaved B, reg-pf)", call_turbo_gemm),
]

results = {}
for name, kernel_func in kernels:
    try:
        print(f"\nTesting {name}...")
        result = kernel_func(A, B)
        results[name] = result
        check_correctness(name, result, ref_result, tol=1e-3)
    except Exception as e:
        print(f"  Ã¢Å“â€” ERROR: {e}")
        results[name] = None

# Test with medium matrices
print("\n" + "="*60)
print("CORRECTNESS TEST: Medium Matrices (256x256x256)")
print("="*60)

m, n, k = 256, 256, 256
A = np.random.randn(m, k).astype(np.float32)
B = np.random.randn(k, n).astype(np.float32)
ref_result = np.dot(A, B).astype(np.float32)

for name, kernel_func in kernels:
    try:
        print(f"\nTesting {name}...")
        result = kernel_func(A, B)
        check_correctness(name, result, ref_result, tol=1e-3)
    except Exception as e:
        print(f"  Ã¢Å“â€” ERROR: {e}")
````

## Section 7: Performance Benchmarking

Benchmark the CUDA kernels against PyTorch matmul and measure GFLOP/s.

````python
import numpy as np
import torch
import time

def benchmark_kernel(name, kernel_func, a, b, warmup=3, iterations=10):
    """Benchmark a GEMM kernel and return GFLOP/s."""
    
    # Warmup runs
    for _ in range(warmup):
        _ = kernel_func(a, b)
    
    # Timed runs
    times = []
    for _ in range(iterations):
        start = time.time()
        result = kernel_func(a, b)
        end = time.time()
        times.append(end - start)
    
    avg_time = np.mean(times[1:])  # Skip first iteration
    std_time = np.std(times[1:])
    
    # Calculate GFLOP/s: 2*m*n*k FLOPs
    m, k = a.shape
    k, n = b.shape
    flops = 2 * m * n * k
    gflops = (flops / 1e9) / avg_time
    
    return {
        'name': name,
        'time_ms': avg_time * 1000,
        'gflops': gflops,
        'std_ms': std_time * 1000,
    }

def benchmark_pytorch(a_np, b_np, warmup=3, iterations=10):
    """Benchmark PyTorch matmul on GPU."""
    a_torch = torch.from_numpy(a_np).cuda()
    b_torch = torch.from_numpy(b_np).cuda()
    
    # Warmup
    for _ in range(warmup):
        _ = torch.matmul(a_torch, b_torch)
    
    torch.cuda.synchronize()
    
    # Timed runs
    times = []
    for _ in range(iterations):
        torch.cuda.synchronize()
        start = time.time()
        result = torch.matmul(a_torch, b_torch)
        torch.cuda.synchronize()
        end = time.time()
        times.append(end - start)
    
    avg_time = np.mean(times[1:])
    std_time = np.std(times[1:])
    
    # Calculate GFLOP/s
    m, k = a_np.shape
    k, n = b_np.shape
    flops = 2 * m * n * k
    gflops = (flops / 1e9) / avg_time
    
    return {
        'name': 'PyTorch matmul (GPU)',
        'time_ms': avg_time * 1000,
        'gflops': gflops,
        'std_ms': std_time * 1000,
    }

# Benchmark with diverse matrix shapes:
#   square        - compute-bound, tests peak throughput
#   tall-skinny   - large m, small n  (e.g. batch matmul, attention)
#   wide-short    - small m, large n  (transpose of tall-skinny)
#   rectangular   - unequal m/n/k     (typical DNN weight shapes)
test_sizes = [
    # (m, n, k, label)
    (512,  512,  512,  "Square 512"),
    (1024, 1024, 1024, "Square 1024"),
    (2048, 2048, 2048, "Square 2048"),
    (4096, 128,  4096, "Tall-Skinny 4096x128"),
    (8192, 64,   2048, "Tall-Skinny 8192x64"),
    (128,  4096, 4096, "Wide-Short 128x4096"),
    (1024, 512,  2048, "Rectangular 1024x512x2048"),
    (2048, 256,  1024, "Rectangular 2048x256x1024"),
]

print("="*80)
print("PERFORMANCE BENCHMARK: Comparing GFLOP/s")
print("="*80)

benchmark_results = []

for m, n, k, label in test_sizes:
    print(f"\n{'='*80}")
    print(f"[{label}]  A({m}x{k}) x B({k}x{n}) = C({m}x{n})")
    print(f"Total FLOPs: {2*m*n*k/1e9:.2f}G")
    print(f"{'='*80}")
    
    # Create test matrices
    A = np.random.randn(m, k).astype(np.float32)
    B = np.random.randn(k, n).astype(np.float32)
    
    results_for_size = []
    
    # Benchmark custom kernels
    for name, kernel_func in kernels:
        try:
            result = benchmark_kernel(name, kernel_func, A, B, warmup=3, iterations=10)
            results_for_size.append(result)
            print(f"{name:30s}: {result['gflops']:8.2f} GFLOP/s (Ã‚Â±{result['std_ms']:.2f} ms, time: {result['time_ms']:.2f} ms)")
        except Exception as e:
            print(f"{name:30s}: FAILED - {e}")
    
    # Benchmark PyTorch
    try:
        result = benchmark_pytorch(A, B, warmup=3, iterations=10)
        results_for_size.append(result)
        print(f"{result['name']:30s}: {result['gflops']:8.2f} GFLOP/s (Ã‚Â±{result['std_ms']:.2f} ms, time: {result['time_ms']:.2f} ms)")
    except Exception as e:
        print(f"PyTorch matmul (GPU): FAILED - {e}")
    
    benchmark_results.append((m, n, k, label, results_for_size))

print("\n" + "="*80)
print("SUMMARY TABLE")
print("="*80)
````

````python
# Create summary table
import pandas as pd

summary_data = []
for size_m, size_n, size_k, label, results in benchmark_results:
    size_str = f"{size_m}x{size_n}x{size_k}"
    for result in results:
        summary_data.append({
            'Matrix Size': size_str,
            'Kernel': result['name'],
            'GFLOP/s': result['gflops'],
            'Time (ms)': result['time_ms'],
            'Std Dev (ms)': result['std_ms']
        })

df = pd.DataFrame(summary_data)
print("\n" + df.to_string(index=False))

# Find PyTorch reference performance
print("\n" + "="*80)
print("PERFORMANCE ANALYSIS")
print("="*80)

for size_m, size_n, size_k, label, results in benchmark_results:
    print(f"\nMatrix Size: {size_m}x{size_n}x{size_k}")
    print("-" * 60)
    
    pytorch_perf = None
    for result in results:
        if 'PyTorch' in result['name']:
            pytorch_perf = result['gflops']
            break
    
    if pytorch_perf:
        for result in results:
            if 'PyTorch' not in result['name']:
                efficiency = (result['gflops'] / pytorch_perf) * 100
                print(f"  {result['name']:30s}: {efficiency:6.1f}% of PyTorch")
    else:
        for result in results:
            print(f"  {result['name']:30s}: {result['gflops']:8.2f} GFLOP/s")
````

## Section 8: Visualization

````python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# Extract data for plotting
kernel_names = set()
size_labels = []
size_perf = {}

for size_m, size_n, size_k, label, results in benchmark_results:
    size_str = f"{size_m}x{size_n}"
    size_labels.append(size_str)
    
    for result in results:
        kernel_names.add(result['name'])
        if result['name'] not in size_perf:
            size_perf[result['name']] = []
        size_perf[result['name']].append(result['gflops'])

kernel_names = sorted(list(kernel_names))

# Create bar chart
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Plot 1: Performance by kernel at different sizes
x = np.arange(len(size_labels))
width = 0.2

for i, kernel in enumerate(kernel_names):
    offset = width * (i - len(kernel_names)/2 + 0.5)
    ax1.bar(x + offset, size_perf[kernel], width, label=kernel, alpha=0.8)

ax1.set_xlabel('Matrix Size', fontsize=12, fontweight='bold')
ax1.set_ylabel('GFLOP/s', fontsize=12, fontweight='bold')
ax1.set_title('GEMM Performance Comparison', fontsize=14, fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(size_labels)
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3)

# Plot 2: Efficiency relative to PyTorch
pytorch_idx = next((i for i, name in enumerate(kernel_names) if 'PyTorch' in name), None)

if pytorch_idx is not None:
    for i, kernel in enumerate(kernel_names):
        if 'PyTorch' not in kernel:
            efficiency = [size_perf[kernel][j] / size_perf[kernel_names[pytorch_idx]][j] * 100 
                         for j in range(len(size_labels))]
            ax2.plot(size_labels, efficiency, marker='o', label=kernel, linewidth=2)

ax2.set_xlabel('Matrix Size', fontsize=12, fontweight='bold')
ax2.set_ylabel('Efficiency (% of PyTorch)', fontsize=12, fontweight='bold')
ax2.set_title('Custom Kernels vs PyTorch matmul', fontsize=14, fontweight='bold')
ax2.legend(fontsize=10)
ax2.grid(True, alpha=0.3)
ax2.axhline(y=100, color='k', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.savefig('/root/gemm_benchmark_results.png', dpi=150, bbox_inches='tight')
print("Ã¢Å“â€œ Saved plot to /root/gemm_benchmark_results.png")
plt.show()
````
